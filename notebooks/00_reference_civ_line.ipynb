# Reference: the CIV line used as our test transition

The test line used in the other notebooks,
`Line(species="Test", wave_rest=1550.0, f=0.1, gamma=2.65e8)`, is a
rounded version of a real CIV transition. This notebook gives its atomic
data and shows the doublet in an observed quasar spectrum.

## Atomic data

CIV is a doublet (1548.2 Å and 1550.8 Å), one of the most observed
absorption features in quasar spectra and a tracer of highly ionised
gas in the circumgalactic/intergalactic medium.

Reference values from Verner, Verner & Ferland 1996, *ADNDT* 64, 1
([arXiv:atom-ph/9604003](https://arxiv.org/abs/atom-ph/9604003)):

| line     | wave_rest (Å) | f      | Gamma (s⁻¹) |
|----------|---------------|--------|-------------|
| CIV 1548 | 1548.1950     | 0.190  | 2.64e8      |
| CIV 1550 | 1550.7700     | 0.0952 | 2.64e8      |

- `wave_rest`: vacuum wavelength at z = 0. The observed line centre is
  `line.wave_rest * (1 + z)`, as in `line_profiles.py`.
- Our test line approximates CIV 1550: `f=0.1` vs 0.0952,
  `gamma=2.65e8` vs 2.64e8 (< 1%).

## Observed example

Quasar J0117+3153 (two sightlines, A/B), with intervening CIV systems
marked in blue (solid: 1548 Å, dotted: 1550 Å). Wavelengths are in the
observed frame, shifted by the redshift of each *absorber*, not of the
quasar.

![Real quasar spectrum with CIV absorption systems marked](../figures/civ_quasar_spectrum_full.png)

Zoom on one doublet:

![Zoomed-in CIV doublet pair](../figures/civ_quasar_spectrum_zoom.png)

The two components have f ratio 2:1, so in the linear regime the
1548 Å line has twice the EW of the 1550 Å one; as the lines saturate
the ratio tends to 1. The doublet ratio is therefore a direct
saturation diagnostic, and a reason for `curve_of_growth.py` to fit both
components together rather than our single test line.

Source: Fig. 1 of [Transverse sizes of CIV absorption systems measured
from multiple QSO sightlines](https://arxiv.org/abs/1007.2457).
