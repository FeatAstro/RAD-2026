## Reality check -- what does this line actually look like?

Our `Line(wave_rest=1550.0, f=0.1, gamma=2.65e8)` is a stand-in for a
real strong UV metal transition, modeled after the **CIV doublet**
(rest wavelengths 1548.2 Å and 1550.8 Å), one of the most commonly
observed intervening absorption features in quasar spectra (used as a
tracer of highly ionised gas in the circumgalactic/intergalactic
medium). ### Where our rest wavelength, f, and Gamma actually come from

The real CIV resonance doublet, from a standard atomic-data reference
used in exactly this kind of absorption-line work (D.A. Verner,
E.M. Verner & G.J. Ferland 1996, *Atomic Data for Permitted Resonance
Lines of Atoms and Ions from H to Si, and S, Ar, Ca and Fe*, Atomic
Data and Nuclear Data Tables, 64, 1 -- [arXiv:atom-ph/9604003](https://arxiv.org/abs/atom-ph/9604003)):

| line       | wave_rest (Å) | f (oscillator strength) | Gamma = A (s⁻¹) |
|------------|---------------|--------------------------|-------------------|
| CIV 1548   | 1548.1950     | 0.190                    | 2.64e8            |
| CIV 1550   | 1550.7700     | 0.0952                   | 2.64e8            |

**"Rest wavelength" here means exactly z=0**: it's the wavelength the
transition would be observed at if the absorbing atom had zero
velocity relative to us -- precisely the
`lam0_obs = line.wave_rest * (1 + z)` convention used throughout
`profiles.py`. These are the kind of numbers a real `linelist.py`
would store per transition (species, wave_rest, f, gamma).

Our test `Line(wave_rest=1550.0, f=0.1, gamma=2.65e8)` is a rounded
stand-in sitting between the two real components (closest to the
1550.77 Å line), not an arbitrary guess: `gamma=2.65e8` matches the
real `A=2.64e8 s^-1` to under 1%, and `f=0.1` sits close to the real
1550-line value (0.0952).

Note the real doublet is *two* lines with a 2:1 oscillator-strength
ratio (visible here as the left component being visibly deeper than
the right) -- our single `Line` object is a simplified stand-in for
one component of that pair, not the full doublet. Worth remembering
once `linelist.py` has real atomic data and `cog.py` starts combining
multiple transitions per species.
